# YaRN: stretching the context window from 4K to 128K

*DeepSeek-V3 technical report, section 4.3. Notebook 7 of 8.*

**TL;DR.** DeepSeek-V3 is pre-trained on sequences of at most 4,096 tokens, then extended to 128K with two short extra training phases of 1,000 steps. The tool is **YaRN**, a rule for rescaling the rotary position embedding. Rotary embeddings turn each pair of channels by an angle that grows with position, and each pair turns at its own speed. Fast pairs have been through many full turns during training, so they can safely be left alone. Slow pairs have not completed a single turn, so a longer context asks them for angles they have never seen. YaRN divides the speed of exactly those slow pairs by a factor $s$, leaves the fast ones, blends in between, and adds a small temperature correction. This notebook derives the rule, checks it against the paper's settings ($s=40$, $\alpha=1$, $\beta=32$), measures what it does to the angles, and tests it on a small language model trained at one length and evaluated at longer ones.

| step | what we do |
|---|---|
| 1 | rotary embeddings recalled: one speed per channel pair |
| 2 | which channels are in trouble at a longer context? |
| 3 | position interpolation and YaRN, with V3's constants |
| 4 | a check of the claims on the angles |
| 5 | a language model trained at length 64, evaluated at 128, 256 and 512 |
| 6 | where V3 applies it: only the decoupled key of MLA |

## 1. Rotary embeddings, one speed per channel pair

Rotary position embedding (RoPE) gives a token at position $p$ its position by **rotating** pairs of channels of its query and key. Channel pair $i$ (of $d/2$ pairs) is rotated by the angle $p\,\theta_i$, with

$$ \theta_i = b^{-2i/d},\qquad b = 10000 . $$

So $\theta_0 = 1$ radian per token (a full turn every 6.3 tokens), and the slowest pair turns almost not at all. The **wavelength** of pair $i$ is $\lambda_i = 2\pi/\theta_i$: the number of tokens for one full turn. Because rotations compose, the attention score between positions $m$ and $n$ depends only on the offset $n-m$; fast pairs encode fine, local offsets, slow pairs encode coarse, distant ones. For V3's decoupled RoPE part (64 channels, section 3 of notebook 1), here are the wavelengths:

```python
d_r = 64
theta = K.rope_inv_freq(d_r)
lam = 2 * math.pi / theta
r_trained = 4096 / lam                                   # full turns completed inside the 4096-token training window
print(f"{'pair':>5s} {'theta':>10s} {'wavelength (tokens)':>20s} {'turns in 4096 tokens':>22s}")
for i in [0, 4, 8, 12, 16, 20, 24, 28, 31]:
    print(f"{i:5d} {float(theta[i]):10.5f} {float(lam[i]):20.1f} {float(r_trained[i]):22.3f}")
print(f"\npairs that complete at least one turn in 4096 tokens: {int((r_trained >= 1).sum())} of {d_r // 2}")
print(f"pairs that complete 32 or more turns                 : {int((r_trained >= 32).sum())} of {d_r // 2}")
check("32 pairs for 64 channels", len(theta) == 32)
check("pair 0 turns once per 2*pi tokens", abs(float(lam[0]) - 2 * math.pi) < 1e-5)
```

## 2. Which channels are in trouble?

During pre-training every position is below 4,096. A pair that makes at least one full turn in that window has seen **every angle** (angles repeat every turn), so a longer context only asks it for angles it already knows. A pair that makes **less than one turn** has only seen angles from 0 up to $4096\,\theta_i$ (a small arc), and a longer context asks for angles beyond that arc: **never seen**. Those are the pairs that break when you simply feed longer sequences. Count them, and see how far outside the trained arc a 128K context goes:

```python
orig_T, new_T = 4096, 131072
max_seen = orig_T * theta                                # largest angle ever used in training, per pair
max_new = new_T * theta                                  # largest angle at 128K positions
unseen = r_trained < 1                                   # pairs that never completed a turn in training
print(f"pairs that never completed a turn in training: {int(unseen.sum())} of 32  (these ask for new angles at long range)")
print(f"for those pairs, the angle at 128K is up to {float((max_new / max_seen)[unseen].max()):.0f}x the largest trained angle")
print(f"the other {int((~unseen).sum())} pairs make >= 1 turn and only repeat angles they know")
fig, ax = plt.subplots(figsize=(8, 3.4))
ax.semilogy(range(32), lam.numpy(), "o-", color=C["blue"], label="wavelength of pair i (tokens)")
ax.axhline(4096, color=C["red"], ls="--", label="training context 4096"); ax.axhline(131072, color=C["orange"], ls="--", label="target context 128K")
ax.set(xlabel="channel pair i", ylabel="wavelength (tokens)", title="Pairs above the red line never complete a turn in training"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
check("some pairs are out-of-range at long context", int(unseen.sum()) > 5)
check("the first pairs repeat known angles", not bool(unseen[0]))
```

## 3. Position interpolation, and YaRN

Two ways to make long positions look like trained ones.

**Position interpolation (PI).** Divide every position by $s = L_{new}/L_{old}$, i.e. divide every $\theta_i$ by $s$. Position 128K then produces the angles that position 4K used to. The slow pairs are back inside their trained arc. The price: the **fast** pairs are slowed down by the same factor, so two neighbouring tokens now differ in angle by $1/s$ of what they did in training. A model that learned to tell adjacent tokens apart from the fast pairs now sees almost identical rotations.

**YaRN** (Peng et al.) treats the pairs differently according to $r_i = L_{old}/\lambda_i$, the number of turns made in the original context:

- $r_i < \alpha$ (few turns): **interpolate**, $\theta_i' = \theta_i / s$;
- $r_i > \beta$ (many turns): **keep**, $\theta_i' = \theta_i$;
- in between: a linear ramp, $\theta_i' = (1-\gamma_i)\,\theta_i/s + \gamma_i\,\theta_i$ with $\gamma_i = \mathrm{clip}\!\big(\frac{r_i-\alpha}{\beta-\alpha}, 0, 1\big)$.

On top of that, YaRN makes attention a little *sharper* to compensate for the larger context: logits are scaled by $(0.1\ln s + 1)^2$, implemented by multiplying both queries and keys by $0.1\ln s + 1$ (paper: $\sqrt{1/t} = 0.1\ln s + 1$). DeepSeek-V3 uses $s = 40$, $\alpha = 1$, $\beta = 32$, a 4096-token original window, and applies it only to the decoupled RoPE key. The kit's function:

```python
from dsv3kit import rope_inv_freq   # the base RoPE speeds, defined in notebook 1

def yarn_inv_freq(dim: int, scale: float, orig_T: int, alpha: float = 1.0, beta: float = 32.0, base: float = 10000.0):
    """YaRN: slow channels are stretched by `scale`, fast channels are left alone, a linear ramp in between.

    r_i = orig_T / wavelength_i = how many full turns channel i makes inside the original context.
        r < alpha  -> divide the frequency by `scale`  (position interpolation)
        r > beta   -> keep the frequency               (no change)
    """
    inv = rope_inv_freq(dim, base)
    r = orig_T / (2 * math.pi / inv)
    gamma = ((r - alpha) / (beta - alpha)).clamp(0, 1)         # 0 = interpolate, 1 = keep
    return (1 - gamma) * inv / scale + gamma * inv, gamma


def yarn_mscale(scale: float) -> float:
    """sqrt(1/t) = 0.1 ln(s) + 1: attention-temperature correction."""
    return 0.1 * math.log(scale) + 1.0
```

```python
s_v3, alpha, beta = 40, 1.0, 32.0
inv_yarn, gamma = yarn_inv_freq(d_r, s_v3, 4096, alpha, beta)
inv_pi = theta / s_v3
print("V3 settings: s = %g, alpha = %g, beta = %g, mscale (sqrt(1/t)) = %.4f   (logits scaled by %.3f)" % (s_v3, alpha, beta, yarn_mscale(s_v3), yarn_mscale(s_v3) ** 2))
print(f"pairs fully interpolated (gamma = 0): {int((gamma == 0).sum())}   kept as is (gamma = 1): {int((gamma == 1).sum())}   blended: {int(((gamma > 0) & (gamma < 1)).sum())}")

fig, ax = plt.subplots(1, 2, figsize=(12, 3.5))
ax[0].semilogy(range(32), theta.numpy(), "o-", color=C["grey"], label="original")
ax[0].semilogy(range(32), inv_pi.numpy(), "s-", color=C["orange"], label=f"position interpolation (/{s_v3})")
ax[0].semilogy(range(32), inv_yarn.numpy(), "^-", color=C["green"], label="YaRN")
ax[0].set(xlabel="channel pair i", ylabel="rotation speed per token", title="New rotation speeds"); ax[0].legend(fontsize=8)
ax[1].plot(range(32), gamma.numpy(), "o-", color=C["purple"]); ax[1].set(xlabel="channel pair i", ylabel="gamma (1 = keep, 0 = interpolate)", title="YaRN's ramp")
plt.tight_layout(); plt.show()
check("YaRN keeps the fastest pair unchanged", torch.allclose(inv_yarn[0], theta[0]))
check("YaRN interpolates the slowest pair by s", torch.allclose(inv_yarn[-1], theta[-1] / s_v3))
check("mscale for s=40 is 0.1 ln 40 + 1 = 1.369", abs(yarn_mscale(40) - 1.36889) < 1e-4)
check("with s = 1 YaRN is plain RoPE", torch.allclose(yarn_inv_freq(d_r, 1, 4096)[0], theta))
```

## 4. Checking the claims about the angles

Three claims we can verify directly.

1. After YaRN with $s=40$, every pair that used to ask for unseen angles now stays **inside its trained arc** even at 128K tokens.
2. PI also achieves that, but **destroys the resolution of the fast pairs**: neighbouring tokens become 40 times more alike in those pairs.
3. YaRN leaves that resolution untouched for the fast pairs.

For claim 2 we measure the rotation between two tokens one position apart. For the pair with speed $\theta$ the angle is $\theta$; use $1-\cos\theta$ as a measure of how distinguishable the two are (0 means identical).

```python
max_new_yarn = new_T * inv_yarn
max_new_pi = new_T * inv_pi
inside_yarn = bool((max_new_yarn[unseen] <= max_seen[unseen] * 1.0001).all())
inside_pi = bool((max_new_pi[unseen] <= max_seen[unseen] * 1.0001).all())
print("unseen pairs inside the trained arc at 128K  --  plain RoPE: False   PI: %s   YaRN: %s" % (inside_pi, inside_yarn))
print("(largest ratio  angle at 128K / largest trained angle:  plain %.1f   PI %.2f   YaRN %.2f)" % (
    float((max_new / max_seen)[unseen].max()), float((max_new_pi / max_seen)[unseen].max()), float((max_new_yarn / max_seen)[unseen].max())))

dist = lambda inv: (1 - torch.cos(inv))                           # distinguishability of adjacent tokens, per pair
fast = torch.arange(0, 8)                                          # the eight fastest pairs
print(f"\nadjacent-token distinguishability (1 - cos), mean over the 8 fastest pairs:  original {float(dist(theta)[fast].mean()):.4f}   PI {float(dist(inv_pi)[fast].mean()):.5f}   YaRN {float(dist(inv_yarn)[fast].mean()):.4f}")
check("plain RoPE leaves the unseen pairs outside their trained arc at 128K", float((max_new / max_seen)[unseen].max()) > 10)
check("PI brings all unseen pairs inside the trained arc", inside_pi)
check("YaRN does too", inside_yarn)
check("PI makes the fast pairs 100x+ less discriminating for neighbouring tokens", float(dist(inv_pi)[fast].mean()) < 0.01 * float(dist(theta)[fast].mean()))
check("YaRN leaves the fast pairs untouched", torch.allclose(dist(inv_yarn)[fast], dist(theta)[fast]))
```

So YaRN does the same job as interpolation on the pairs that need it, and does not pay the price on the pairs that don't. The remaining question is whether a model trained at a short length really behaves better at a longer one with the rule applied. That is empirical, so we test it.

## 5. A language model trained at 64, evaluated at 128, 256 and 512

We train the small character model with standard RoPE on sequences of 64 characters (our "4K"), then evaluate it on much longer windows with three position treatments: plain RoPE (just continue to larger positions), PI, and YaRN, each with $s = L_{eval}/64$. We measure the loss **only on positions beyond 64**, the ones the model never saw during training. Then, following the paper, we also fine-tune briefly at the longer length with each treatment and measure again.

The toy has a short window and a small head size (24 channels, 12 pairs), so the YaRN constants of V3 ($\alpha = 1$, $\beta = 32$) classify the pairs differently than at 4K. We keep them as they are.

```python
data = K.CharData()
T0 = 64
cfg = K.LMConfig(vocab=data.vocab, d=96, n_layers=3, n_heads=4, d_ff=192)
t0 = time.time()
model0, _ = K.train_lm(cfg, data, steps=300, B=48, T=T0, lr=4e-3, seed=0)
print("trained at T=%d in %.0f s;  val loss at T=64: %.3f" % (T0, time.time() - t0, K.evaluate(model0, data, T=T0, n_batches=10)["main"]))
```

```python
import copy

def set_rope(model, inv_freq=None, mscale=1.0):
    for b in model.blocks:
        r = b.attn.rope
        r.inv_freq = (K.rope_inv_freq(r.dim) if inv_freq is None else inv_freq.clone())
        r.mscale = mscale

def variant(model, kind, s):
    m = copy.deepcopy(model)
    dim = m.blocks[0].attn.rope.dim
    if kind == "plain": set_rope(m)
    elif kind == "PI": set_rope(m, K.rope_inv_freq(dim) / s)
    elif kind == "YaRN":
        inv, _ = K.yarn_inv_freq(dim, s, T0, alpha=1.0, beta=32.0)
        set_rope(m, inv, K.yarn_mscale(s))
    return m

@torch.no_grad()
def long_loss(model, T_eval, n=12, seed=3):
    """mean next-token loss over positions >= T0 only, on windows of length T_eval"""
    model.eval()
    gen = torch.Generator().manual_seed(seed)
    tot, cnt = 0.0, 0
    for _ in range(n):
        seq = data.batch("val", 4, T_eval, gen)
        logits = model(seq[:, :-1])
        l = F.cross_entropy(logits.reshape(-1, logits.shape[-1]), seq[:, 1:].reshape(-1), reduction="none").view(4, T_eval)
        tot += float(l[:, T0:].sum()); cnt += l[:, T0:].numel()
    return tot / cnt

lens = [128, 256, 512]
tab = {}
print(f"{'':8s}" + "".join(f"{'T=' + str(L):>10s}" for L in lens) + "   (loss on positions beyond 64)")
for kind in ["plain", "PI", "YaRN"]:
    tab[kind] = [long_loss(variant(model0, kind, L / T0), L) for L in lens]
    print(f"{kind:8s}" + "".join(f"{v:10.3f}" for v in tab[kind]))
print(f"{'(T=64)':8s} reference loss inside the trained window: {K.evaluate(model0, data, T=T0, n_batches=10)['main']:.3f}")
```

```python
check("evaluation at longer windows is worse than inside the trained window for plain RoPE", tab["plain"][-1] > K.evaluate(model0, data, T=T0, n_batches=10)["main"])
best = {L: min(tab, key=lambda k: tab[k][i]) for i, L in enumerate(lens)}
print("best treatment at each length (no fine-tuning):", best)
```

Read the table directly. Whether YaRN or PI beats plain extrapolation here depends on how far the model is pushed (the ratio $L_{eval}/64$) and on this model's reliance on position at all: a character model of this size has mostly **local** context, so very long positions matter little to its predictions. If plain RoPE is already competitive, that is the toy telling us it has little long-range structure to lose, not that the method is unnecessary.

Now the paper's second ingredient: a short fine-tune at the new length. We continue training for 100 steps on windows of 256 characters, once with each treatment active, and evaluate again at 256 and 512.

```python
def finetune(model, kind, s, steps=100, T=256, B=8, lr=1e-3, seed=11):
    m = variant(model, kind, s)
    m.train()
    opt = torch.optim.AdamW(m.parameters(), lr=lr, betas=(0.9, 0.95), weight_decay=0.1)
    gen = torch.Generator().manual_seed(seed)
    for _ in range(steps):
        loss, _ = K.lm_loss(m, data.batch("train", B, T, gen))
        opt.zero_grad(); loss.backward(); torch.nn.utils.clip_grad_norm_(m.parameters(), 1.0); opt.step()
    return m

t0 = time.time()
ft = {}
s_ft = 256 / T0
for kind in ["plain", "PI", "YaRN"]:
    m = finetune(model0, kind, s_ft)
    ft[kind] = [long_loss(m, 256), long_loss(m, 512)]
    print(f"{kind:6s} fine-tuned at 256:  loss beyond 64 at T=256 {ft[kind][0]:.3f}   at T=512 {ft[kind][1]:.3f}")
print("time %.0f s" % (time.time() - t0))
check("fine-tuning at the longer length helps plain RoPE at 256", ft["plain"][0] < tab["plain"][1])
```

Again compare against the differences between treatments, not just their order: with a single trained model and a single fine-tune per treatment we have no seed spread, so small differences are not evidence. We report what we measured and no more.

## 6. Where V3 applies it: only the decoupled key

In DeepSeek-V3's attention (notebook 1) the **content** parts of queries and keys carry no position at all. Position enters only through the decoupled RoPE query and key ($\mathbf q^R, \mathbf k^R$, 64 channels). That is also the only place YaRN is applied (paper: "applied exclusively to the decoupled shared key $\mathbf k^R_t$"). A consequence we can verify with the kit's MLA layer: if the RoPE query projection is zeroed, the layer has no position-dependent term and changing the rotary frequencies must leave the output **exactly** unchanged.

```python
torch.manual_seed(0)
mla = K.MLA(d=64, n_h=4, d_h=16, d_c=12, d_cq=20, d_r=8).eval()
h = torch.randn(2, 20, 64)
with torch.no_grad():
    base_out = mla(h)
    mla.rope.inv_freq = yarn_inv_freq(8, 8, 8)[0]; mla.rope.mscale = yarn_mscale(8)     # YaRN on the RoPE part
    yarn_out = mla(h)
    changed = float((base_out - yarn_out).abs().max())
    mla.W_QR.weight.zero_()                                                              # remove the positional term from the scores
    mla.rope.inv_freq = K.rope_inv_freq(8); mla.rope.mscale = 1.0
    a = mla(h)
    mla.rope.inv_freq = yarn_inv_freq(8, 8, 8)[0]; mla.rope.mscale = yarn_mscale(8)
    b = mla(h)
print(f"changing the rotary frequencies moves the output by {changed:.3f}")
print(f"with the RoPE query zeroed, the same change moves it by {float((a - b).abs().max()):.2e}")
check("YaRN on the decoupled part changes the output", changed > 1e-3)
check("without a RoPE query the content path is position-free, so the output does not move", float((a - b).abs().max()) < 1e-6)
```

## What to take away

1. **RoPE gives each channel pair its own rotation speed.** Pairs that complete a full turn in the training window have seen every angle; pairs that don't will be asked for unseen angles at longer context.
2. **YaRN divides the speed of the slow pairs by $s$, leaves the fast ones alone and blends in between**, with a logit-sharpening factor $(0.1\ln s + 1)^2$. We verified on the angles that this keeps every previously-unseen pair inside its trained arc at 128K (like interpolation) without flattening the fast pairs (unlike interpolation).
3. **V3's recipe:** $s=40$, $\alpha=1$, $\beta=32$, on the decoupled RoPE key only, then two fine-tuning phases of 1,000 steps at 32K and 128K.

## What this notebook does not show
- **Long-context quality at V3 scale.** The paper's evidence is the "needle in a haystack" test up to 128K (its Figure 8). Ours is a 64-character-trained toy evaluated up to 512, whose predictions depend little on distant context. We report the numbers as measured; they do not test whether YaRN preserves retrieval.
- **A rigorous comparison.** One trained model, one fine-tune per treatment, no seeds: differences between plain, PI and YaRN in the tables are indicative at best.
- **A win for YaRN.** In our run plain RoPE was the best treatment at every length without fine-tuning, and after fine-tuning at 256 it was best at 256; YaRN was ahead of plain only at 512 (2.544 against 2.560), a margin we cannot call real. Our model is local (characters), so it has little long-range structure to preserve.
- **The 32K and 128K training phases** (batch sizes 1920 and 480 with a learning rate of $7.3\times10^{-6}$): we ran 100 toy steps.
